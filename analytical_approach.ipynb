```python
import numpy as np

def expected_strokes_one_hole(transition_matrix, transient_states, absorbing_states):
    """
    Compute expected number of strokes to absorption for one hole.
    
    Parameters
    ----------
    transition_matrix : np.ndarray
        Full transition probability matrix (square).
    transient_states : list[int]
        Indices of transient states (e.g., tee, fairway, green).
    absorbing_states : list[int]
        Indices of absorbing states (e.g., cup).
        
    Returns
    -------
    expected_strokes : float
        Expected number of strokes to complete one hole.
    """
    # Extract Q submatrix (transient -> transient)
    Q = transition_matrix[np.ix_(transient_states, transient_states)]
    
    # Fundamental matrix: N = (I - Q)^(-1)
    I = np.eye(Q.shape[0])
    N = np.linalg.inv(I - Q)
    
    # Expected number of steps to absorption = sum of each row of N
    # Assuming we start at the first transient state (e.g., tee)
    t = N.sum(axis=1)
    
    return t[0]  # expected strokes starting from "tee"


def expected_strokes_round(transition_matrix, transient_states, absorbing_states, n_holes=18):
    """
    Compute expected total strokes for n_holes holes.
    """
    per_hole = expected_strokes_one_hole(transition_matrix, transient_states, absorbing_states)
    return per_hole * n_holes


if __name__ == "__main__":
    # Example transition matrix
    # Order: [tee, fairway, green, cup]
    # Cup is absorbing (100% stay in cup)
    P = np.array([
        [0.0, 0.6, 0.3, 0.1],  # from tee
        [0.0, 0.5, 0.3, 0.2],  # from fairway
        [0.0, 0.0, 0.4, 0.6],  # from green
        [0.0, 0.0, 0.0, 1.0],  # from cup (absorbing)
    ])
    
    transient = [0, 1, 2]   # tee, fairway, green
    absorbing = [3]         # cup
    
    one_hole = expected_strokes_one_hole(P, transient, absorbing)
    full_round = expected_strokes_round(P, transient, absorbing, n_holes=18)
    
    print(f"Expected strokes per hole: {one_hole:.3f}")
    print(f"Expected strokes for 18 holes: {full_round:.3f}")
```
